```python
import string
import re
from pathlib import Path
```

### Todas as letras com capitalização minúscula

```python
input_path = Path('carolina_output/')
output_path = Path('test_min/')

output_path.mkdir(exist_ok=True) 

input_files = [file for file in input_path.glob('*.txt')]
input_files = sorted(input_files, key=lambda f: f.stat().st_size)

for file in input_files:
    out_file = output_path / f'{file.stem}_min{file.suffix}'
    with open(file, 'r', encoding='utf-8') as inf, \
         open(out_file, 'w', encoding='utf-8') as outf:
        print(f"[{file}] Criando teste lowercase...")
        for line in inf:
            lowercase = line.lower()

            outf.write(lowercase)
        print(f"[{file}] Teste lowercase criado!")
```

### Todas as letras com capitalização minúscula e pontuação removida

```python
input_path = Path('carolina_output/')
output_path = Path('test_min_nopunct/')

output_path.mkdir(exist_ok=True) 

input_files = [file for file in input_path.glob('*.txt')]
input_files = sorted(input_files, key=lambda f: f.stat().st_size)

regex = re.compile(r'[^\w\s]|_')
for file in input_files:
    out_file = output_path / f'{file.stem}_min_nopunct{file.suffix}'
    with open(file, 'r', encoding='utf-8') as inf, \
         open(out_file, 'w', encoding='utf-8') as outf:
        print(f"[{file}] Criando teste lowercase sem pontuação...")
        for line in inf:
            final_line = line.lower()
            translator = str.maketrans('', '', string.punctuation)
            final_line = regex.sub('', final_line)
            outf.write(final_line)
        print(f"[{file}] Teste lowercase sem pontuação criado!")
```

### Todas as letras com capitalização minúscula e sem acentos

```python
input_path = Path('carolina_output/')
output_path = Path('test_min_noaccent/')

output_path.mkdir(exist_ok=True) 

input_files = [file for file in input_path.glob('*.txt')]
input_files = sorted(input_files, key=lambda f: f.stat().st_size)

com_acentos = "áàâãäéèêëíìîïóòôõöúùûüçÁÀÂÃÄÉÈÊËÍÌÎÏÓÒÔÕÖÚÙÛÜÇ"
sem_acentos = "aaaaaeeeeiiiiooooouuuucAAAAAEEEEIIIIOOOOOUUUUC"

for file in input_files:
    out_file = output_path / f'{file.stem}_min_noaccent{file.suffix}'
    with open(file, 'r', encoding='utf-8') as inf, \
         open(out_file, 'w', encoding='utf-8') as outf:
        print(f"[{file}] Criando teste lowercase sem acentos ...")
        for line in inf:
            final_line = line.lower()

            translator = str.maketrans(com_acentos, sem_acentos)
            final_line = final_line.translate(translator)

            outf.write(final_line)
        print(f"[{file}] Teste lowercase sem acentos criado!")
             
```

```python
input_path = Path('carolina_output/')
output_path = Path('test_min_noaccent_nopunct/')

output_path.mkdir(exist_ok=True) 

input_files = [file for file in input_path.glob('*.txt')]
input_files = sorted(input_files, key=lambda f: f.stat().st_size)

com_acentos = "áàâãäéèêëíìîïóòôõöúùûüçÁÀÂÃÄÉÈÊËÍÌÎÏÓÒÔÕÖÚÙÛÜÇ"
sem_acentos = "aaaaaeeeeiiiiooooouuuucAAAAAEEEEIIIIOOOOOUUUUC"

regex = re.compile(r'[^\w\s]|_')
translator = str.maketrans(com_acentos, sem_acentos)
for file in input_files:
    out_file = output_path / f'{file.stem}_min_noaccent_nopunct{file.suffix}'
    with open(file, 'r', encoding='utf-8') as inf, \
         open(out_file, 'w', encoding='utf-8') as outf:
        print(f"[{file}] Criando teste lowercase sem pontuação e sem acentos ...")
        for line in inf:
            final_line = line.lower()

            final_line = final_line.translate(translator)

            final_line = regex.sub('', final_line)
            
            outf.write(final_line)
        print(f"[{file}] Teste lowercase sem pontuação e sem acentos criado!")
             
```
